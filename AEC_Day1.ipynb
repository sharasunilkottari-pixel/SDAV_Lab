```python
import numpy as np
marks=np.array([65,72,81,55,90])
print("Mean:",marks.mean())
print("Maximum:",marks.max())
print("Minimum:",marks.min())
print("Std. Dev:",marks.std())
```

```python
import pandas as pd
data={
    "Name": ["A","B","C","D","E"],
    "Marks":[65,72,81,55,90]
}
df=pd.DataFrame(data)
print(df)
print(df["Marks"].mean())
```

```python
import matplotlib.pyplot as plt
plt.bar(df["Name"],df["Marks"])
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
plt.hist(df["Marks"])
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Marks")
plt.show()
```

# **Experiment 1:Probablity through simulation**
np.random.choice() randomly slects outcomes.Size=10 means 10 trials.
On running the cell again  and agian we get different result everytime

```python
result=np.random.choice(["Head","Tail"],size=10)
print(result)
```

```python
result=np.random.choice(["Head","Tail"],size=10)
heads=np.sum(result=="Head")
print("Heads:",heads)
print("Probality of Heads:",heads/10)
```

```python
result=np.random.choice(["Head","Tail"],size=100)
heads=np.sum(result=="Head")
print("Heads:",heads)
print("Probality of Heads:",heads/100)
```

```python
result=np.random.choice(["Head","Tail"],size=1000)
heads=np.sum(result=="Head")
print("Heads:",heads)
print("Probality of Heads:",heads/1000)
```

```python
result=np.random.choice(["Head","Tail"],size=10000)
heads=np.sum(result=="Head")
print("Heads:",heads)
print("Probality of Heads:",heads/10000)
```

**Visulize the Coin Toss**

```python
counts=[
    np.sum(result=="Head"),
    np.sum(result=="Tail")
]
plt.bar(["Head","Tail"],counts)
plt.xlabel("Outcome")
plt.ylabel("Count")
plt.title("Coin Toss Simulation")
plt.show()
```

```python
trial=[10,100,1000,10000]
for n in trial:
  result=np.random.choice(["Head","Tail"],size=n)
  probabilty=np.sum(result=="Head")/n
  print(n,probabilty)
```

Rolling a Die


```python
trial=[10,100,1000,10000]
for n in trial:
  result=np.random.choice([1,2,3,4,5,6],size=n)
  probabilty=np.sum(result==3)/n
  print(n,probabilty)
```

```python
num=int(input("Enter the number you want to check probablity for:"))
trial=int(input("Enter number of trial:"))
result=np.random.choice([1,2,3,4,5,6],size=trial)
probabilty=np.sum(result==num)/trial
print("The probality of ",num," is ",probabilty)
```

```python
counts=[
    np.sum(result==1),
    np.sum(result==2),
    np.sum(result==3),
    np.sum(result==4),
    np.sum(result==5),
    np.sum(result==6),
]
plt.bar([1,2,3,4,5,6],counts)
plt.xlabel("Outcome")
plt.ylabel("Count")
plt.title("Dice Toss Simulation")
plt.show()
```
