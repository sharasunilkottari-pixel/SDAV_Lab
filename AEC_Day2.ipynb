**Question3:**
A die is rolled 1000 times find experimental conditional probablity P(getting even number|number is greater than 3)

```python
import numpy as np
result=np.random.choice([1,2,3,4,5,6],size=1000)
p1=0 #number is even and greater than 3
p2=0 #number greater than 3
for i in range(1,7):
  if i%2==0 and i>3:
    p1+=(np.sum(result==i)/1000)
  if  i>3:
    p2+=(np.sum(result==i)/1000)
print("P(getting even number|number is greater than 3) : ",p1/p2)
```
