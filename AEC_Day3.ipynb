2. Dice Probability
Write a Python program to calculate the probability of getting:
- a) An even number
- b) An odd number
- c) A number greater than 4
- d) A number less than or equal to 3
Use the sample space of a single die.

```python
die=list(range(1,7))
#print("Sample Space: ",die)
p1=len([x for x in die if x%2==0])/len(die)
p2=len([x for x in die if x%2!=0])/len(die)
p3=len([x for x in die if x>4])/len(die)
p4=len([x for x in die if x<=3])/len(die)
print("P(An even number):",p1)
print("P(An odd number):",p2)
print("P(A number greater than 4):",p3)
print("P(A number less than or equal to 3):",p4)
```

**3**. Two-Dice Experiment
Generate the sample space for rolling two dice and calculate the probability that:
- a) The sum is 7
- b) The sum is greater than 8
- c) Both dice show the same number
- d) At least one die shows 6
Hint: Use itertools.product(range(1,7), range(1,7)).

```python
die=list(range(1,7))
twodie=list(itertools.product(die,die))
#print("Sample Space: ",twodie)
p1=len([x for x in twodie if sum(x)==7])/len(twodie)
p2=len([x for x in twodie if sum(x)>8])/len(twodie)
p3=len([x for x in twodie if x[0]==x[1]])/len(twodie)
p4=len([x for x in twodie if 6 in x])/len(twodie)
print("P(The sum is 7)",p1)
print("P(The sum is greater than 8)",p2)
print("P(Both dice show the same number)",p3)
print("P( At least one die shows 6)",p4)
```

5. Law of Large Numbers
Simulate 10,000 fair coin tosses and calculate the cumulative probability of Heads after every toss.
Plot:
- X-axis → Number of tosses
- Y-axis → Cumulative probability of Heads
- Horizontal line → \(P(H)=0.5\)
Use a logarithmic X-axis.

```python
n = 10000
tosses  = np.random.binomial(1, 0.5, n)
#
emp_prob = [tosses[:i].mean() for i in range(1, n+1)]
plt.figure(figsize=(10,4))
plt.plot(emp_prob, color='steelblue', linewidth=1)
plt.axhline(0.5, color='red', linestyle='--', label='True P=0.5')
plt.xscale('log')
#
plt.title('Empirical Probability Convergence (Law of Large Numbers)')
plt.xlabel('Number of tosses')
plt.ylabel('Cumulative probability of Heads')
plt.legend(); plt.tight_layout(); plt.show()
```

6. Conditional Probability with Dice
Roll a die.
Define:
- \(A\): number is even
- \(B\): number is greater than 3
- Write a program to calculate: P(A|B)
- Verify the result manually.

```python
die=list(range(1,7))
#print("Sample Space: ",die)
a=[x for x in die if x%2==0]
b=[x for x in die if x>3]
#pa=len(a)/len(die)
pb=len(b)/len(die)
pab=len(set(a)&set(b))/len(die)
p=pab/pb
print("P(number is even|number is greater than 3):",p)
```

7. Bayes' Theorem - Medical Test Problem
A disease affects 2% of a population.
A diagnostic test has:
- Sensitivity = 95%
- False-positive rate = 4%
- Write a Python program to calculate: P(Disease|Positive)
- Display the answer as a percentage.

```python
disease=0.02
positive=0.95
negative=0.04

totalpositive=positive*disease+negative*(1-disease)
prob=(positive*disease)/totalpositive
print(f" P(Disease|Positive) :{prob:.4f}")
print(f"Percentage: {prob*100:.2f}%")
```

8. Email Classification
- Suppose: 20% of emails are spam.
- A spam email contains the word "offer" with probability 80%.
- A non-spam email contains offer with probability 10%.
- Using Bayes theorem, calculate: P(Spam|offer)
- Write a Python program to display the result.

```python
spam=0.2
offerspam=0.8
offernospam=0.1
total=spam*offerspam+(1-spam)*offernospam
required=spam*offerspam/total
print(f"P(Spam|offer): {required:.4f}")
print(f"Percentage: {required*100:.2f}%")
```
